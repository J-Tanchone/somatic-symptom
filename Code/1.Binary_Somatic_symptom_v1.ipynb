## Variables and Measures

### Outcome Variable

The outcome variable were assessed using item 13 from the **Patient Health Questionnaire (PHQ)**. Responses were recoded into a binary variable, where a score of 2 or higher was considered symptomatic (coded as 1), and a score of 1 indicated no significant fatigue symptoms (coded as 0). This binary formulation was chosen due to its moderate class balance and clinical relevance.

### Psychological Predictors

All psychological variables were computed as mean scores across item-level responses, provided that a sufficient proportion of items (≥70%) were present. The following scales were included:

- **IDEA-8**: Identity exploration and development (8 items)
- **Markers of Adulthood**:
  - Achievement subscale (10 items)
  - Importance subscale (10 items)
- **Perceived Stress Scale** (10 items)
- **Perceived Social Support** (12 items)
- **Need to Belong Scale** (10 items)
- **Mindfulness Scale** (15 items)
- **Self-Efficacy/Competence** (10 items)
- **Narcissistic Personality Inventory (NPI-13)** (13 items)
- **Interpersonal Exploitativeness** (3 items)
- **Disability Identity and Status** (combination of 22 items from Q10, Q11, and Q14)
- **Social Media Use**:
  - Maintaining connections (5 items)
  - Making new connections (4 items)
  - Seeking information (2 items)
- **Subjective Well-Being** (6 items)
- **Interpersonal Transgressions** (4 items)
- **Belief in and importance of the American Dream** (2 items)

### Demographic and Attitudinal Predictors

- **Sex** (male, female, other; categorical)
- **Education level**
- **Race/Ethnicity**
- **Household income**
- **School attended**
- **Parental marriage status** (recoded into categorical dummy variables)
- **Siblings** (recoded: -0.5 = no siblings, 0.5 = at least one sibling)
- **Importance of marriage** (single-item rating)

These variables represent a theoretically grounded and diverse set of predictors for modeling somatic symptom risk, with particular focus on psychological and social functioning within a young adult population.

## Split Data

```python
# Get physSx_ column names
physSx_cols = [col for col in final_data.columns if col.startswith('physSx_')]
# Create X by dropping all physSx_ columns
X = final_data.drop(columns=physSx_cols, errors='ignore')

# Store all train/test splits
train_test_data = {}

# Loop through each physSx_ variable as dependent variable
for physSx_var in physSx_cols:
    print(f"\nPreparing data for: {physSx_var}")

    # Create y for current physSx_ variable
    y = data[physSx_var]

    # Combine X and y, drop missing values
    model_data = X.join(y.rename(physSx_var)).dropna()

    # Separate X and y for modeling
    X_clean = model_data.drop(columns=[physSx_var])
    y_clean = model_data[physSx_var]

    # Train-test split
    X_train, X_test, y_train, y_test = train_test_split(
        X_clean, y_clean,
        test_size=0.3,
        random_state=45,
        stratify=y_clean
    )

    # Store the splits
    train_test_data[physSx_var] = {
        'X_train': X_train,
        'X_test': X_test,
        'y_train': y_train,
        'y_test': y_test,
        'X_clean': X_clean
    }

    print(f"Train shape: {X_train.shape}")
    print(f"Test shape: {X_test.shape}")
    print("Target distribution:")
    print(y_clean.value_counts(normalize=True))

print(f"\nData preparation complete for {len(train_test_data)} physSx variables")
```

## Model Accuracy and Comparison

```python
# Define features once
numeric_features = [col for col in X.columns if col.endswith('_m') or col.endswith('_c')]
categorical_features = [col for col in ['sex', 'edu', 'race', 'income', 'parental_marriage'] if col in X.columns]
categorical_features = [col for col in categorical_features if col in X_clean.columns]

# Convert categorical variables to string (important!)
for col in categorical_features:
  X_train[col] = X_train[col].astype(str)
  X_test[col] = X_test[col].astype(str)

# Preprocessing pipelines
scaled_preprocessor = ColumnTransformer(transformers=[
    ('num', StandardScaler(), numeric_features),
    ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
])

unscaled_preprocessor = ColumnTransformer(transformers=[
    ('num', 'passthrough', numeric_features),
    ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
])

# Updated models with class_weight='balanced'
pipelines = {
    'Logistic Regression': Pipeline([
        ('preprocessor', scaled_preprocessor),
        ('classifier', LogisticRegression(max_iter=1000, class_weight='balanced'))
    ]),
    'KNN': Pipeline([
        ('preprocessor', scaled_preprocessor),
        ('classifier', KNeighborsClassifier(n_neighbors=5))
    ]),
    'Random Forest': Pipeline([
        ('preprocessor', unscaled_preprocessor),
        ('classifier', RandomForestClassifier(
            n_estimators=100,
            class_weight='balanced',
            random_state=42
        ))
    ])
}
```

```python
# Loop through each physSx_ variable for modeling
for physSx_var, data_splits in train_test_data.items():
    print(f"\n{'='*60}")
    print(f"Running models for: {physSx_var}")
    print(f"{'='*60}")
    # Fit & evaluate
    for name, pipeline in pipelines.items():
        pipeline.fit(X_train, y_train)
        y_pred = pipeline.predict(X_test)
        acc = accuracy_score(y_test, y_pred)
        print(f"\n{name} Accuracy: {acc:.3f}")
        print(classification_report(y_test, y_pred))
```

## Logistic Regression Model

```python
# Initialize results storage
results_list = []
fitted_models = {}

# Define the regularization types to test
regularization_types = {
    'ElasticNet': {'penalty': 'elasticnet', 'solver': 'saga', 'l1_ratio': [0.1, 0.3, 0.5, 0.7, 0.9]},
    'LASSO': {'penalty': 'l1', 'solver': 'saga', 'l1_ratio': None},
    'Ridge': {'penalty': 'l2', 'solver': 'lbfgs', 'l1_ratio': None}
}

for physSx_var, data_splits in train_test_data.items():
    # Extract the data splits for this specific physSx variable
    X_train = data_splits['X_train'].copy()
    X_test = data_splits['X_test'].copy()
    y_train = data_splits['y_train']
    y_test = data_splits['y_test']

    # Convert categorical variables to string
    for col in categorical_features:
        X_train[col] = X_train[col].astype(str)
        X_test[col] = X_test[col].astype(str)

    # Test each regularization type
    for reg_name, reg_params in regularization_types.items():
        # Build parameter grid
        param_grid = {'classifier__C': [0.01, 0.1, 1, 10]}

        # Add l1_ratio for Elastic Net
        if reg_params['l1_ratio'] is not None:
            param_grid['classifier__l1_ratio'] = reg_params['l1_ratio']

        # Create the pipeline
        pipeline = Pipeline([
            ('preprocessor', scaled_preprocessor),
            ('classifier', LogisticRegression(
                max_iter=1000,
                class_weight='balanced',
                penalty=reg_params['penalty'],
                solver=reg_params['solver']
            ))
        ])

        # Perform grid search
        grid = GridSearchCV(
            pipeline,
            param_grid=param_grid,
            cv=5,
            scoring='f1'
        )

        grid.fit(X_train, y_train)

        # Store the fitted model with a unique key
        model_key = f"{physSx_var}_{reg_name}"
        fitted_models[model_key] = grid

        # Prepare results dictionary
        result_dict = {
            'Variable': physSx_var,
            'Regularization': reg_name,
            'Best_C': grid.best_params_['classifier__C'],
            'Best_F1_Score': round(grid.best_score_, 3)
        }

        # Add l1_ratio if it's Elastic Net
        if 'classifier__l1_ratio' in grid.best_params_:
            result_dict['Best_l1_ratio'] = grid.best_params_['classifier__l1_ratio']
        else:
            result_dict['Best_l1_ratio'] = None

        results_list.append(result_dict)

# Create DataFrame after the loop
lr_results = pd.DataFrame(results_list)
lr_results = lr_results.sort_values('Best_F1_Score', ascending=False)

# Display the results
print("\nLogistic Regression Results with Different Regularization Types:")
print("="*80)
display(lr_results)
```

```python
# === Neural Net with pure Keras + manual CV (robust; no SciKeras needed) ===
import numpy as np
from itertools import product
from collections import defaultdict

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (roc_auc_score, average_precision_score,
                             balanced_accuracy_score, classification_report, confusion_matrix)
from sklearn.utils.class_weight import compute_class_weight
from sklearn.pipeline import Pipeline  # only for preprocessing convenience

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, callbacks, regularizers

# Reproducibility
tf.keras.utils.set_random_seed(42)
np.random.seed(42)

def _dense_ohe():
    try:
        return OneHotEncoder(handle_unknown="ignore", sparse_output=False)  # sklearn >=1.2
    except TypeError:
        return OneHotEncoder(handle_unknown="ignore", sparse=False)         # older sklearn

def make_preprocessor(numeric_features, categorical_features):
    # ColumnTransformer that outputs dense arrays
    return ColumnTransformer(
        transformers=[
            ("num", Pipeline([
                ("impute", SimpleImputer(strategy="median")),
                ("scale",  StandardScaler())
            ]), numeric_features),
            ("cat", Pipeline([
                ("impute", SimpleImputer(strategy="most_frequent")),
                ("oh",     _dense_ohe())
            ]), categorical_features),
        ],
        remainder="drop"
    )

def build_binary_mlp(input_dim, n_units=64, n_layers=2, dropout=0.2, lr=1e-3, l2=1e-4):
    model = keras.Sequential([layers.Input(shape=(input_dim,))])
    for _ in range(n_layers):
        model.add(layers.Dense(n_units, activation="relu",
                               kernel_regularizer=regularizers.l2(l2)))
        if dropout > 0:
            model.add(layers.Dropout(dropout))
    model.add(layers.Dense(1, activation="sigmoid"))
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=lr),
        loss="binary_crossentropy",
        metrics=[keras.metrics.AUC(name="auc"),
                 keras.metrics.Precision(name="precision"),
                 keras.metrics.Recall(name="recall")]
    )
    return model

es = callbacks.EarlyStopping(monitor="val_auc", mode="max", patience=6, restore_best_weights=True)

# Manual hyperparameter grid
grid = {
    "n_units":  [64, 128],
    "n_layers": [1, 2],
    "dropout":  [0.0, 0.3],
    "lr":       [1e-3, 3e-4],
    "l2":       [1e-5, 1e-4],
}
grid_list = [dict(zip(grid.keys(), vals)) for vals in product(*grid.values())]

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Ensure results collector exists
try:
    all_results
except NameError:
    all_results = []

# ================== Loop per outcome ==================
for physSx_var, splits in train_test_data.items():
    X_train_df = splits["X_train"].copy()
    X_test_df  = splits["X_test"].copy()
    y_train = splits["y_train"].astype(int).to_numpy()
    y_test  = splits["y_test"].astype(int).to_numpy()

    # Feature lists (adjust to your schema)
    numeric_features = [c for c in X_train_df.columns if c.endswith("_m") or c.endswith("_c")]
    categorical_features = [c for c in ['sex','edu','race','income','parental_marriage'] if c in X_train_df.columns]

    # Ensure categoricals are strings
    for col in categorical_features:
        X_train_df[col] = X_train_df[col].astype(str)
        X_test_df[col]  = X_test_df[col].astype(str)

    # Class weights for imbalance
    classes = np.array([0, 1])
    cw = compute_class_weight(class_weight="balanced", classes=classes, y=y_train)
    class_weight_dict = {int(k): float(v) for k, v in zip(classes, cw)}
    print(f"\n{'='*60}\nOutcome: {physSx_var} | Class weights: {class_weight_dict}")

    # --------- Manual CV over grid (ROC AUC as selector; switch to PR-AUC if you prefer) ---------
    best_mean, best_std, best_params = -np.inf, np.inf, None

    for params in grid_list:
        fold_scores = []

        for tr_idx, va_idx in cv.split(X_train_df, y_train):
            X_tr_df, X_va_df = X_train_df.iloc[tr_idx], X_train_df.iloc[va_idx]
            y_tr, y_va = y_train[tr_idx], y_train[va_idx]

            # Fit preprocessor on the fold's train; transform both train and val
            pre = make_preprocessor(numeric_features, categorical_features)
            X_tr = pre.fit_transform(X_tr_df)
            X_va = pre.transform(X_va_df)

            model = build_binary_mlp(
                input_dim=X_tr.shape[1],
                n_units=params["n_units"],
                n_layers=params["n_layers"],
                dropout=params["dropout"],
                lr=params["lr"],
                l2=params["l2"]
            )

            history = model.fit(
                X_tr, y_tr,
                epochs=50,
                batch_size=64,
                verbose=0,
                class_weight=class_weight_dict,
                validation_split=0.1,  # ensures val_auc for early stopping
                callbacks=[es],
                shuffle=True
            )

            y_va_prob = model.predict(X_va, verbose=0).ravel()
            fold_scores.append(roc_auc_score(y_va, y_va_prob))

        mean_auc = float(np.mean(fold_scores))
        std_auc  = float(np.std(fold_scores))
        if mean_auc > best_mean:
            best_mean, best_std, best_params = mean_auc, std_auc, params

    print(f"Best CV ROC AUC: {best_mean:.3f} ± {best_std:.3f} with params: {best_params}")

    # --------- Refit best on full training and evaluate on test ---------
    pre_best = make_preprocessor(numeric_features, categorical_features)
    X_train = pre_best.fit_transform(X_train_df)
    X_test  = pre_best.transform(X_test_df)

    model_best = build_binary_mlp(
        input_dim=X_train.shape[1],
        n_units=best_params["n_units"],
        n_layers=best_params["n_layers"],
        dropout=best_params["dropout"],
        lr=best_params["lr"],
        l2=best_params["l2"]
    )

    model_best.fit(
        X_train, y_train,
        epochs=50,
        batch_size=64,
        verbose=0,
        class_weight=class_weight_dict,
        validation_split=0.1,
        callbacks=[es],
        shuffle=True
    )

    y_prob = model_best.predict(X_test, verbose=0).ravel()
    y_pred = (y_prob >= 0.5).astype(int)

    auc  = roc_auc_score(y_test, y_prob)
    ap   = average_precision_score(y_test, y_prob)
    bal  = balanced_accuracy_score(y_test, y_pred)

    print(f"\nNeural Net (manual CV best) {best_params}")
    print(f"AUC: {auc:.3f} | PR-AUC: {ap:.3f} | Balanced Acc: {bal:.3f}")
    print(classification_report(y_test, y_pred, digits=3))
    print("Confusion matrix:\n", confusion_matrix(y_test, y_pred))

    all_results.append({
        "outcome": physSx_var,
        "model": f"Neural Net (manual CV) {best_params}",
        "AUC": auc,
        "PR_AUC": ap,
        "BalancedAcc": bal
    })

print("\n✅ Neural Net manual-CV block finished successfully (no SciKeras).")
```

```python
# ====== SHAP INTERPRETABILITY FOR THE NN ======
import shap, os, numpy as np
import matplotlib.pyplot as plt

# 0) Make a folder per-outcome for artifacts
out_dir = f"shap_nn_{physSx_var}"
os.makedirs(out_dir, exist_ok=True)

# 1) Get feature names from the fitted preprocessor
def _feature_names_from_ct(ct, input_cols):
    """Return expanded feature names after ColumnTransformer fit."""
    names = []
    for name, trans, cols in ct.transformers_:
        if name == "remainder" and trans == "drop":
            continue
        if hasattr(trans, "get_feature_names_out"):
            # Pipeline -> last step may have names; fall back if needed
            try:
                fn = trans.get_feature_names_out()
                fn = list(fn)
            except:
                # try last step
                if hasattr(trans, "steps") and hasattr(trans.steps[-1][1], "get_feature_names_out"):
                    fn = list(trans.steps[-1][1].get_feature_names_out())
                else:
                    fn = list(np.array(cols, dtype=str))
        else:
            # Plain transformer; just repeat original col names
            fn = list(np.array(cols, dtype=str))
        # If OneHotEncoder, names already expanded; if StandardScaler, names unchanged
        names.extend(fn)
    return names

feature_names = _feature_names_from_ct(pre_best, X_train_df.columns)

# 2) Pick a small background set for SHAP (for speed) and a test sample to explain
rng = np.random.default_rng(42)
bg_idx = rng.choice(X_train.shape[0], size=min(200, X_train.shape[0]), replace=False)
X_bg   = X_train[bg_idx]

# explain up to 300 test rows to keep runtime reasonable
k_explain = min(300, X_test.shape[0])
ex_idx    = rng.choice(X_test.shape[0], size=k_explain, replace=False)
X_te_s    = X_test[ex_idx]
y_te_s    = y_test[ex_idx]

# 3) Build SHAP explainer
#    shap.Explainer auto-selects (TFGradient/Deep/Kernel). If it fails, fallback to KernelExplainer.
try:
    explainer = shap.Explainer(model_best, X_bg)  # model outputs probabilities (sigmoid)
    shap_values = explainer(X_te_s)
except Exception as e:
    print(f"[SHAP] Fast explainer failed ({e}). Falling back to KernelExplainer (slower).")
    f = lambda xx: model_best.predict(xx, verbose=0).ravel()
    explainer = shap.KernelExplainer(f, X_bg)
    shap_values = explainer.shap_values(X_te_s, nsamples="auto")  # returns array

# 4) Normalize outputs to a consistent object
# shap.Explanation → has .values and .feature_names; KernelExplainer may return np.array
if isinstance(shap_values, shap.Explanation):
    sv = shap_values.values  # shape: (n_samples, n_features)
else:
    sv = np.array(shap_values)  # (n_samples, n_features)

# 5) Global importance: mean |SHAP|
mean_abs = np.abs(sv).mean(axis=0)
order = np.argsort(-mean_abs)
topk = min(25, len(feature_names))
imp_table = np.c_[mean_abs[order][:topk], order[:topk]]
# Save CSV with all features
import pandas as pd
df_imp = pd.DataFrame({
    "feature": np.array(feature_names)[order],
    "mean_abs_shap": mean_abs[order]
})
df_imp.to_csv(os.path.join(out_dir, f"shap_mean_abs_{physSx_var}.csv"), index=False)

# 6) Plots — bar (top-25) and beeswarm (top-25)
plt.figure(figsize=(8, 6))
plt.barh(np.array(feature_names)[order[:topk]][::-1], mean_abs[order[:topk]][::-1])
plt.title(f"{physSx_var} — NN SHAP (mean |SHAP|, top {topk})")
plt.tight_layout()
plt.savefig(os.path.join(out_dir, f"{physSx_var}_shap_bar_top{topk}.png"), dpi=150)
plt.close()

# beeswarm: need shap.Explanation with data for feature values; build it if needed
try:
    if isinstance(shap_values, shap.Explanation):
        shap.plots.beeswarm(shap_values[:, order[:topk]], max_display=topk, show=False)
    else:
        # Build an Explanation wrapper so beeswarm shows value colors
        shap_exp = shap.Explanation(values=sv, data=X_te_s, feature_names=feature_names)
        shap.plots.beeswarm(shap_exp[:, order[:topk]], max_display=topk, show=False)
    plt.title(f"{physSx_var} — NN SHAP beeswarm (top {topk})")
    plt.tight_layout()
    plt.savefig(os.path.join(out_dir, f"{physSx_var}_shap_beeswarm_top{topk}.png"), dpi=150)
    plt.close()
except Exception as e:
    print(f"[SHAP] Beeswarm failed: {e}")

# 7) One local force plot (first explained sample)
try:
    if isinstance(shap_values, shap.Explanation):
        shap.plots.force(shap_values[0], matplotlib=True, show=False)
    else:
        shap_exp = shap.Explanation(values=sv, data=X_te_s, feature_names=feature_names)
        shap.plots.force(shap_exp[0], matplotlib=True, show=False)
    plt.title(f"{physSx_var} — NN SHAP force (example 0)")
    plt.tight_layout()
    plt.savefig(os.path.join(out_dir, f"{physSx_var}_shap_force_example0.png"), dpi=150)
    plt.close()
except Exception as e:
    print(f"[SHAP] Force plot failed: {e}")

print(f"[SHAP] Saved to: {out_dir}/  "
      f"Files: shap_mean_abs_{physSx_var}.csv, "
      f"{physSx_var}_shap_bar_top{topk}.png, "
      f"{physSx_var}_shap_beeswarm_top{topk}.png, "
      f"{physSx_var}_shap_force_example0.png")
# ====== END SHAP ======
```
